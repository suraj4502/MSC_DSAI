```python
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
import requests as re
```

```python
url = 'https://www.worldometers.info/world-population/population-by-country/'
src = re.get(url)
print("Status Code::" , src.status_code)
```

```python
soup = BeautifulSoup(src.content, features="lxml")
print(soup.prettify())
```

```python
soup.body()
```

```python
t1 = soup.find('table')
t1
```

```python
cols = []
t_header = t1.find('thead')
t_header

for tr in t_header.find_all('tr'):
    cols.append(tr.text)
    
print(cols)
```

```python
t_body = t1.find('tbody')
t_rows = t_body.find_all('tr')
d1 = []

for rows in t_rows:
    data = []
    for row_data in rows.find_all('td'):
        data.append(row_data.text)
    d1.append(data)
    
d1
```

```python
df = pd.DataFrame(d1)
df.head()
```

 ### data extraction from AN XML file

```python
import pandas as pd
import xml.etree.ElementTree as et
```

```python
x_tree = et.parse('student_data.xml')
x_tree
```

```python
root = x_tree.getroot()
root.tag
```

```python
rows = []
for student in root.findall('student'):
    # Extract data from child elements
    sname = student.find('Name').text if student.find('Name') is not None else None
    semail = student.find('Email').text if student.find('Email') is not None else None
    srollno = student.find('RollNo').text if student.find('RollNo') is not None else None
    sage = student.find('Age').text if student.find('Age') is not None else None

    # Append the extracted data to the rows list
    rows.append({
        'Name': sname,
        'Email': semail,
        'RollNo': srollno,
        'Age': sage
    })
```

```python
df=pd.DataFrame(rows)
df
```
