# single layer Perceptron.

```python
import numpy as np
```

```python
np.zeros(5)
```

```python
class Perceptron:
    def __init__(self, input_size, learining_rate = 0.1, epochs = 10):
        # iniitalize weights and bias
        self.weight = np.zeros(input_size)
        self.bias = 0.0
        self.learining_rate = learining_rate
        self.epochs = epochs
        
    def step(self, x):
        # Step activation function
        return 1 if x >= 0 else 0
   
    def predict(self, x):
        # linear combination
        linear_output = np.dot(self.weight, x) + self.bias
        return self.step(linear_output)
    
    def train(self, X, y):
        """
        X: numpy array of shape (num_samples, input_size)
        y: numpy array of shape (num_samples,) with labels 0 or 1
        """
        for _ in range(self.epochs):
            for i in range(len(X)):
                # make prediction
                prediction = self.predict(X[i])
                
                # update weights
                error = y[i] - prediction
                if error != 0:
                    self.weight += self.learining_rate *error * X[i]
                    self.bias += self.learining_rate * error
        
```

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0,0,0,1])

perceptron = Perceptron(input_size=2,
                       learining_rate=0.2,
                       epochs=10)
perceptron.train(X,y)
```

```python
test_inputs = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

for inp in test_inputs:
    print(f"Input :: {inp}, predited Output :: {perceptron.predict(inp)}")
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
import numpy as np

X = np.array([[0,0],[0,1],[1,0],[1,1]], dtype=np.float32)
y = np.array([[0],[0],[0],[1]], dtype=np.float32)

model = Sequential([
    Dense(1, activation='sigmoid', input_shape=(2,))
])
model.compile(optimizer='sgd', loss='binary_crossentropy', metrics=['accuracy'])
model.fit(X, y, epochs=100, verbose=0)
loss, acc = model.evaluate(X, y, verbose=0)
print("Loss:", loss, "Accuracy:", acc)
predictions = model.predict(X)
print(predictions)
```

# 2. Implementation of Multi-Layer Perceptron.

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, accuracy_score

import numpy as np
import matplotlib.pyplot as plt
```

```python
data = load_breast_cancer()
X = data.data
y = data.target
```

```python
# split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=452)
```

```python
# Scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# MODEL BUILDING

model = Sequential([
    Dense(64, activation= 'relu', input_shape = (X_train.shape[1],)),
    Dense(32, activation= 'relu'),
    Dense(1, activation= 'sigmoid')
])

model.compile(optimizer= 'adam', loss='binary_crossentropy', metrics=['accuracy'])

# Model Training
history = model.fit(X_train, y_train, verbose=1)

loss, accuracy = model.evaluate(X_test, y_test, verbose=1)
print(f"Test Accuracy :: {accuracy:.2f}")
```

```python
y_pred = (model.predict(X_test) > 0.5).astype("int32")
print(classification_report(y_test, y_pred))
```
