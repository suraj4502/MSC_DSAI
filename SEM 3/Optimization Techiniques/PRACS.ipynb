# golden method

```python
#golden method
import math
def golden(f, a, b, tol= 1e-2):
    phi = (1 + 5 ** 0.5)/2
    print("PHI ::", phi)
    
    while b-a > tol:
        c = b - (b-a) / phi
        d = a + (b-a) / phi
        
        if f(c) < f(d):
            b = d
        else:
            a = c
            
    return (a+b)/2

f = lambda x: (x-3)**2 + 2

print(golden(f, 0,6))
```

# Steepest Descent on quardratinc equation

```python
import numpy as np

# Corresponds to grad = A*x - B for f = (x-1)^2 + 4(y-2)^2

A = np.diag([2.0, 8.0])
B = np.array([2.0, 16.0])
x = np.array([0.0, 0.0])

for k in range(5):
    g = A @ x - B
    print("G ::", g )
    
    alpha = (g @ g)/ (g@(A@g))
    print("ALPHA ::", alpha)
    
    x = x-alpha * g
    print("x ::", x)
    
    print(f"\nstep {k} :: x = {x}")
    
```

```python
# newton method


def newton_minimize(f1, f2, x0, iters = 5):
    x = x0
    for _ in range(iters):
        x = x - f1(x)/ f2(x)
    return x

f1 = lambda x : 4 * x ** 3 - 9 * x ** 2
f2 = lambda x : 12 * x ** 2 - 18 * x

print(newton_minimize(f1, f2, 3))
```

```python
from scipy.optimize import minimize

def objective(x):
    return x[0]**2 + x[1]**2

def constraint1(x):
    return x[0] + x[1] - 1

bounds = ((0, None), (0, None))
constraints = ({'type': 'ineq', 'fun': constraint1})
x0 = [0.5, 0.5]

result = minimize(objective,
                  x0,
                  method = "SLSQP", # use 'method', not 'type'
                  bounds = bounds,
                  constraints = constraints # use 'constraints', not 'constraint'
                 )

print("optimal x:", result.x)
print("optimal value:", result.fun)
print("success:", result.success)
print("message:", result.message)
```

```python
# INEQUALITY CONSTRAINt

from scipy.optimize import minimize

def objective(x):
    return x[0]**2 + x[1]**2

def constraint1(x):
    return x[0] + x[1] - 1

bounds = ((0, None), (0, None))

constraints = ({'type': 'ineq',
               'fun': constraint1})
print("CONSTRAINTS ::", constraints)

x0 = [0.5, 0.5]

result = minimize(objective,
                  x0,
                  type = "SLSQP", 
                  bounds = bounds,
                  constraint = constraints
                 )

# Print the results
print("optimal x:", result.x)
print("optimal value:", result.fun)
print("success:", result.success)
print("message:", result.message)
```

```python
# Quadratic Interpolation Method
import numpy as np

# Objective function: Has a minimum at x = 2.5, f(x) = 3
f = lambda x: (x - 2.5) ** 2 + 3

def quadratic_interpolation(f, x0, x1, x2, tol=1e-5, max_iter=20):
    for k in range(max_iter):
        # Evaluate function at three points
        f0, f1, f2 = f(x0), f(x1), f(x2)

        # Formula for vertex (minimum) of a parabola through three points
        numerator = ((x1**2 - x2**2) * f0 +
                     (x2**2 - x0**2) * f1 +
                     (x0**2 - x1**2) * f2)
        denominator = ((x1 - x2) * f0 +
                       (x2 - x0) * f1 +
                       (x0 - x1) * f2)

        if denominator == 0:
            break

        x_min = 0.5 * numerator / denominator
        f_min = f(x_min)

        print(f"Iter {k+1}: x_min={x_min:.6f}, f(x)={f_min:.6f}")

        # Replace the point with highest function value with new estimate
        if f0 > f1 and f0 > f2:
            x0 = x_min
        elif f1 > f0 and f1 > f2:
            x1 = x_min
        else:
            x2 = x_min

        # Stopping criterion
        if max(abs(x_min - x0), abs(x_min - x1), abs(x_min - x2)) < tol:
            return x_min, f_min

    return x_min, f_min

# Run method with initial guesses 0, 3, 5
x_opt, f_opt = quadratic_interpolation(f, 0, 3, 5)
print("\nQuadratic Interpolation Optimum: x =", round(x_opt, 5), "f(x)=", round(f_opt, 5))
```

```python
# Unconstrained optimization : Elimination Method 
# Interval halving Method -  
import math

# Define the function you want to minimize
f = lambda x: (x - 2.5)**2 + 3  # Minimum at x=2.5, f(x)=3

def interval_halving(f, a, b, tol=1e-3):
    k = 0
    while (b - a) > tol:
        k += 1
        mid = (a + b) / 2
        l = (a + mid) / 2
        r = (mid + b) / 2

        f_l, f_mid, f_r = f(l), f(mid), f(r)

        # Elimination logic
        if f_l < f_mid:
            b = mid
        elif f_r < f_mid:
            a = mid
        else:
            a, b = l, r

        print(f"Iter {k}: Interval=({a:.4f}, {b:.4f})")
    return (a + b) / 2

x_opt = interval_halving(f, 0, 6)
print("Interval Halving Optimum:", x_opt, "f(x)=", f(x_opt))
```
