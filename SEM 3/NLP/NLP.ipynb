```python
import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://www.scrapethissite.com/pages/simple'

response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")
```

```python
countries=soup.find_all("div", class_ = "country")
```

```python
countries[0]
```

```python
data = []

for country in countries:
    name = country.find("h3", class_ ="country-name").get_text(strip=True)
    capital = country.find("span", class_="country-capital" ).get_text(strip=True)
    population = country.find("span", class_ = "country-population").get_text(strip=True)
    area = country.find("span", class_ = "country-area").get_text(strip=True)
    data.append([name, capital, population, area])
    print(name, "->",capital, "->", area)
```

```python
df = pd.DataFrame(data, columns = ['name', 'capital', 'population', 'area'])
df
```

```python
df.to_csv("output.csv", index = False)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import string
import pandas as pd

# Download NLTK tokenizer package (run once)
nltk.download('punkt')

# Sample Corpus
corpus = [
    "Python is amazing and fun.",
    "Python is not just fun but also powerful.",
    "Learning Python is fun!",
]

# Preprocessing Function
def preprocess(text):
    # Lowercase
    text = text.lower()
    # Remove punctuation
    text = text.translate(str.maketrans("", "", string.punctuation))
    # Tokenize into words
    tokens = word_tokenize(text)
    return tokens

# Process the corpus
processed_corpus = [preprocess(sentence) for sentence in corpus]

# Build vocabulary (unique words across corpus)
vocabulary = sorted(set(word for sentence in processed_corpus for word in sentence))
print("Vocabulary:", vocabulary)
print("Vocabulary size:", len(vocabulary))

# Function to create BoW vector for a sentence
def create_bow_vector(sentence, vocab):
    vector = [0] * len(vocab)
    for word in sentence:
        if word in vocab:
            idx = vocab.index(word)
            vector[idx] += 1
    return vector

# Create BoW vectors for all sentences
bow_vectors = [create_bow_vector(sentence, vocabulary) for sentence in processed_corpus]

# Print BoW vectors
print("Bag of Words Vectors:")
for vector in bow_vectors:
    print(vector)
```

# NGRAMS

An n-gram is a sequence of “n” items (usually words or characters) from a given text.

Unigram = 1 word sequence (e.g., “hello”, “world”)

Bigram = 2 word sequence (e.g., “hello world”)

Trigram = 3 word sequence (e.g., “the quick brown”)

N-grams are used in language modeling, text analysis, and many NLP tasks. They capture word patterns and context better than using single words alone.

```python
import pandas as pd
import nltk
from nltk.util import ngrams
from nltk.corpus import reuters
from nltk import trigrams
from collections import defaultdict

nltk.download("punkt")
nltk.download("reuters")

```

```python
df = pd.DataFrame({
    "word": ["apple", "banana", "cherry", "date"],
    "POS": ["pronoun", "verb", "noun", "adverb"]
})
```

```python

df
```

```python
one_hot = pd.get_dummies(df["word"], prefix="token")
print(one_hot)
```

```python
def generate_ngrams(tokens, n):
    return [tokens[i:i+n] for i in range(len(tokens)-n+1)]
```

```python
sentence = "I love Python programming".split()
print(generate_ngrams(sentence, 1))   # Unigrams
print(generate_ngrams(sentence, 2))   # Bigrams
print(generate_ngrams(sentence, 3))   # Trigrams
```

```python
words = [w.lower() for w in reuters.words() if w.isalpha()]
tri_grams = list(trigrams(words))

model = defaultdict(lambda: defaultdict(lambda: 0))
for w1, w2, w3 in tri_grams:
    model[(w1, w2)][w3] += 1
# Calculate probabilities
for w1_w2 in model:
    total = float(sum(model[w1_w2].values()))
    for w3 in model[w1_w2]:
        model[w1_w2][w3] /= total
```

```python
def predict_next_word(w1, w2):
    next_word_probs = model[w1, w2]
    if next_word_probs:
        return max(next_word_probs, key=next_word_probs.get)
    return "No prediction available"

print(predict_next_word("the", "market"))
```

```python
data = {
    "id": [1, 2, 3],
    "text": [
        "I love natural language processing",
        "Natural language processing is fun",
        "I love deep learning"
    ]
}
df = pd.DataFrame(data)

def create_ngrams(text, n):
    tokens = text.lower().split()
    return list(ngrams(tokens, n))

df["bigrams"] = df["text"].apply(lambda x: create_ngrams(x, 2))
df["trigrams"] = df["text"].apply(lambda x: create_ngrams(x, 3))

df[["text", "bigrams", "trigrams"]]
```

# TF-IDF -->  Term Frequency - Inverse Document Frequency

TF-IDF is used to convert text documents into numeric feature vectors that give higher scores to words that are unique or rare in a document but not common everywhere. This helps in many NLP tasks like classification or clustering.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
documents = [
    "The cat sat on the mat.",
    "The dog sat on the log.",
    "The cat chased the mouse.",
    "The dog barked at the cat.",
    "Cats and dogs are common pets."
]
```

```python
vectorizer = TfidfVectorizer(stop_words='english')
```

```python
tfidf_matrix = vectorizer.fit_transform(documents)
```

```python
df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=vectorizer.get_feature_names_out())
df_tfidf
```

# Word embeddings
Word embeddings are dense, multi-dimensional numeric representations of words.
They are learned from real text so that words with similar meaning or use have similar vector values (positions in the multi-dimensional space).
Unlike one-hot encoding or Bag of Words, word embeddings capture semantic meaning and the relationships between words, making them very useful for machine learning and NLP applications.
Semantic meaning: Capture similarity of words (“dog” and “cat” are closer than “dog” and “mat”).

Reduces dimensionality: Each word is a compact vector, not a huge sparse array.

Input for deep learning: Vectors can be fed into complex models for sentiment analysis, translation, etc.

```python
from gensim.models import Word2Vec

sentences = [
    ['the', 'cat', 'sat', 'on', 'the', 'mat'],
    ['the', 'dog', 'sat', 'on', 'the', 'rug'],
    ['the', 'cat', 'chased', 'the', 'mouse'],
    ['the', 'dog', 'barked', 'at', 'the', 'cat']
]

model = Word2Vec(sentences, vector_size=50, window=3, min_count=1, sg=1)
```

```python
vector_cat = model.wv['cat']
print("Vector for 'cat':\n", vector_cat)
```

```python
similar_to_cat = model.wv.most_similar('cat', topn=7)
for word, score in similar_to_cat:
    print(f"word: {word}, score: {score:.4f}")
```
